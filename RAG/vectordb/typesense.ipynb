- We have to create a collection where we can store different documents
- For collection, we first have to define schema

```python
books_schema = {
    'name': 'books',
    'fields': [
        {'name': 'title', 'type': 'string'},
        {'name': 'authors', 'type': 'string[]', 'facet': True},
        {'name': 'publication_year', 'type': 'int32', 'facet': True},
        {'name': 'ratings_count', 'type': 'int32'},
        {'name': 'average_rating', 'type': 'float'}
    ],
    'default_sorting_field': 'ratings_count'
}
```

```python
client.collections.create(books_schema)
```

# Data Import

```python
with open('books.jsonl','r',encoding='utf-8') as f:
    data=f.read()
client.collections['books'].documents.import_(data)
```

# Search in Typesense cloud

```python
search_params={
    'q':"harry potter",
    'query_by':'title,authors',
    'sort_by':'ratings_count:desc'
}
```

```python
client.collections['books'].documents.search(search_parameters=search_params)
```

# RAG + Typesense

```python
from langchain_community.document_loaders import TextLoader
from langchain_community.vectorstores import Typesense 
from langchain_text_splitters import CharacterTextSplitter
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_groq import ChatGroq
from dotenv import load_dotenv
import os
load_dotenv()
```

```python
loader = TextLoader("test.txt")
documents = loader.load()
text_splitter = CharacterTextSplitter(chunk_size=1000, chunk_overlap=100)
docs = text_splitter.split_documents(documents)
embeddings = HuggingFaceEmbeddings()
```

```python
docsearch=Typesense.from_documents(
    docs,
    embeddings,
    typesense_client_params={
    
        'host':'xg6y5atuoeh3mwisp-1.a1.typesense.net',     # We get this info from cluster we are creating
        'port':'443',       #Use this to connect to typesense cloud
        'protocol':'https',
        'typesense_api_key':os.getenv('TYPESENSE_API_KEY'),
        'connection_timeout_seconds':2,
        'typesense_collection_name':"Lang-Chain"
    
   
}
)
```

# LangChain uses that model to calculate the embeddings behind the scenes:
LangChain asks for the embedding object in the configuration for two specific automation tasks:
- Automated -Ingestion (add_texts): When you pass plain text documents to LangChain to store, LangChain uses the object to automatically convert those texts into numbers before sending the payload to Typesense. 
-Automated Querying (similarity_search): When you search for a text phrase like "How to setup search?", LangChain seamlessly converts that text into a vector first using your model, and then asks Typesense to look for nearby vector matches.

```python
query="What is harry Potter?"
res=docsearch.similarity_search(query=query)  #	Returns typesense documents most similar to query.
for doc in res:
    print("Data: ")
    print(doc.page_content)
```
